# Data Preprocessing

This notebook prepares the Telco Customer Churn dataset for machine learning.

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv")

df.head()
```

```python
df["TotalCharges"].dtype
```

```python
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)
```

```python
df["TotalCharges"].isna().sum()
```

```python
df["TotalCharges"] = df["TotalCharges"].fillna(0)
```

```python
print("Missing values:", df["TotalCharges"].isna().sum())
print("Data type:", df["TotalCharges"].dtype)
```

### TotalCharges

`TotalCharges` was initially stored as a string because 11 newly acquired
customers (`tenure = 0`) had blank values.

The feature was converted to numeric format, and these missing values were
replaced with 0, since these customers had not yet accumulated total charges.

## Feature and Target Separation

```python
x = df.drop(columns=["customerID", "Churn"])
y = df["Churn"]
```

```python
x.shape
```

```python
y.shape
```

```python
y.value_counts()
```

```python
y = y.map({
    "No": 0,
    "Yes": 1
})
```

```python
y.value_counts()
```

```python
x.dtypes
```

```python
numeric_features = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

categorical_features = [
    col for col in x.columns
    if col not in numeric_features
]
```

```python
print("Numerical features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

```python
numeric_transformer = StandardScaler()
```

```python
categorical_transformer = OneHotEncoder(
    handle_unknown="ignore"
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("Train:", x_train.shape)
print("Test:", x_test.shape)
```

```python
print("Train churn distribution:")
print(y_train.value_counts(normalize=True))

print("\nTest churn distribution:")
print(y_test.value_counts(normalize=True))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
```

```python
logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", LogisticRegression(
            max_iter=1000,
            random_state=42
        ))
    ]
)
```

```python
logistic_model.fit(x_train, y_train)

y_pred = logistic_model.predict(x_test)
```

### Baseline Model Comparison

The baseline Logistic Regression outperformed the baseline Random Forest
across all evaluated metrics.

In particular, Logistic Regression achieved a higher recall for the churn
class, detecting more customers who actually churned.

Therefore, model complexity does not provide an improvement at this stage.
Further model selection and tuning should be performed using cross-validation
on the training set before final evaluation on the test set.

```python
from sklearn.model_selection import StratifiedKFold, cross_validate

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
scoring = {
    "accuracy": "accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1",
    "roc_auc": "roc_auc"
}

cv_logistic = cross_validate(
    logistic_model,
    x_train,
    y_train,
    cv=cv,
    scoring=scoring
)
```

```python
for metric in scoring:
    scores = cv_logistic[f"test_{metric}"]

    print(
        f"{metric}: "
        f"{scores.mean():.3f} +/- {scores.std():.3f}"
    )
```

```python
cv_rf = cross_validate(
    random_forest_model,
    x_train,
    y_train,
    cv=cv,
    scoring=scoring
)

for metric in scoring:
    scores = cv_rf[f"test_{metric}"]

    print(
        f"{metric}: "
        f"{scores.mean():.3f} +/- {scores.std():.3f}"
    )
```

### Cross-Validation Conclusion

Logistic Regression outperformed the baseline Random Forest across all
evaluation metrics.

Logistic Regression achieved a mean ROC-AUC of 0.846 and a mean recall
of 0.544, compared with 0.820 and 0.482 respectively for Random Forest.

Therefore, Logistic Regression is retained as the current baseline model.
However, recall remains relatively low, meaning that many churners are
still not detected. The next experiments will focus on improving churn
detection while controlling the precision-recall trade-off.

```python
from sklearn.model_selection import cross_val_predict
```

```python
y_train_proba = cross_val_predict(
    logistic_model,
    x_train,
    y_train,
    cv=cv,
    method="predict_proba"
)[:, 1]
```

```python
import pandas as pd

thresholds = [0.50, 0.45, 0.40, 0.35, 0.30]

results = []

for threshold in thresholds:

    y_pred_threshold = (y_train_proba >= threshold).astype(int)

    results.append({
        "Threshold": threshold,
        "Precision": precision_score(y_train, y_pred_threshold),
        "Recall": recall_score(y_train, y_pred_threshold),
        "F1": f1_score(y_train, y_pred_threshold)
    })

threshold_results = pd.DataFrame(results)

threshold_results
```

### Classification Threshold Analysis

Lowering the classification threshold increased recall at the cost of
precision.

At the default threshold of 0.50, Logistic Regression achieved a recall
of approximately 54.4%. Reducing the threshold to 0.35 increased recall
to approximately 71.7%, while precision decreased from 65.2% to 57.1%.

Among the evaluated thresholds, 0.35 achieved the highest F1-score
(approximately 0.635), suggesting a better precision-recall balance.

However, the final threshold should ultimately depend on the business
costs associated with false positives and false negatives.

## Logistic Regression with Class Weight

```python
balanced_logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", LogisticRegression(
            max_iter=1000,
            class_weight="balanced",
            random_state=42
        ))
    ]
)
```

```python
cv_balanced_logistic = cross_validate(
    balanced_logistic_model,
    x_train,
    y_train,
    cv=cv,
    scoring=scoring
)

for metric in scoring:
    scores = cv_balanced_logistic[f"test_{metric}"]

    print(
        f"{metric}: "
        f"{scores.mean():.3f} +/- {scores.std():.3f}"
    )
```

### Impact of Class Weighting

Using `class_weight="balanced"` substantially improved the model's ability
to detect churners.

The mean recall increased from approximately 0.544 to 0.801, while the
F1-score improved from 0.592 to 0.628.

This improvement came at the cost of lower precision (0.517) and accuracy
(0.748), indicating a higher number of false-positive churn alerts.

ROC-AUC remained approximately unchanged at 0.846, suggesting that class
weighting mainly changed the classification behavior rather than the
overall ranking ability of the model.

For a churn prevention use case where missing a churner is considered more
costly than contacting a non-churner, the balanced model may therefore be
more appropriate.

### Model Selection

Cross-validation results show that no single model dominates every metric.

The standard Logistic Regression provides higher accuracy and precision,
while the balanced Logistic Regression substantially improves churn recall
from 0.544 to 0.801 and also achieves the highest F1-score.

Both Logistic Regression variants achieve a similar ROC-AUC of approximately
0.846, while the baseline Random Forest performs worse overall.

Since the primary business objective is to identify customers at risk of
churning, the balanced Logistic Regression is retained as the main candidate
for further optimization.

```python
param_grid = {
    "classifier__C": [0.01, 0.1, 1, 10, 100],
    "classifier__solver": ["liblinear", "lbfgs"]
}
```

```python
from sklearn.model_selection import GridSearchCV

grid_search = GridSearchCV(
    estimator=balanced_logistic_model,
    param_grid=param_grid,
    scoring="f1",
    cv=cv,
    n_jobs=-1,
    verbose=1
)
```

```python
grid_search.fit(x_train, y_train)
```

```python
print("Best parameters:", grid_search.best_params_)
print("Best CV F1:", grid_search.best_score_)
```

```python
best_logistic_model = grid_search.best_estimator_
```

### Hyperparameter Tuning Conclusion

GridSearchCV selected a balanced Logistic Regression with `C=0.1`
and the `liblinear` solver.

The best mean cross-validation F1-score was approximately 0.629.

However, the differences between the tested configurations were small,
indicating that model performance is relatively stable across the

Hyperparameter tuning therefore provided only a marginal improvement
over the initial balanced Logistic Regression.
